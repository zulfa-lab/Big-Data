```python
from pymongo import MongoClient

# Inisialisasi client dan koneksi ke database
client = MongoClient('mongodb://localhost:27017/')
db = client['company_db']
collection = db['employees']

# Contoh untuk memastikan koneksi
print('Connected to MongoDB')
```

```python
# Contoh Insert Data
employee_data = {
    'name': 'Alice',
    'department': 'Finance',
    'age': 29,
    'salary': 4500
}
collection.insert_one(employee_data)
print('Data inserted')

# Tugas: Insert beberapa data tambahan, lakukan update, serta delete data

# Data tambahan
new_employees = [
    {'name': 'Bob', 'department': 'IT', 'age': 32, 'salary': 5200},
    {'name': 'Charlie', 'department': 'HR', 'age': 27, 'salary': 4100},
    {'name': 'Diana', 'department': 'Finance', 'age': 35, 'salary': 6000},
    {'name': 'Eve', 'department': 'Marketing', 'age': 30, 'salary': 4800}
]

# Insert multiple data
collection.insert_many(new_employees)
print("Data tambahan berhasil diinsert!")
```

```python
# Menampilkan semua data yang ada
for employee in collection.find():
    print(employee)
```

```python
# Menampilkan hanya karyawan departemen IT
for emp in collection.find({'department': 'IT'}):
    print(emp)
# Menampilkan karyawan dengan gaji lebih dari 5000
for emp in collection.find({'salary': {'$gt': 5000}}):
    print(emp)
# Menampilkan data tertentu saja (nama dan gaji):
for emp in collection.find({}, {'_id': 0, 'name': 1, 'salary': 1}):
    print(emp)
```

```python
# Update data (gaji satu departemen naik 10%)
collection.update_many(
    {'department': 'Finance'},
     {'$mul': {'salary': 1.10}} #naik 10%
)
print("Gaji karyawan Finance berhasil diperbarui")

# Update data mengubah usia "Bob"
collection.update_one(
    {'name': 'Bob'},
    {'$set': {'age': 33}}
)
print("Data Bob diperbarui")

# HapusCharlie
collection.delete_one({'name': 'Charlie'})
print("Data Charlie dihapus!")

collection.delete_many({'department': 'Marketing'})
print("Semua data Marketing dihapus!")


```

```python
# Latihan Tambahan
# Latihan 4.1: Buatlah koleksi baru products dan masukkan data produk (minimal 10 produk).
# Latihan 4.2: Lakukan query untuk menemukan produk dengan harga di atas rata-rata.
# Latihan 4.3: Buatlah aggregation pipeline untuk menghitung total produk dalam setiap kategori

from pymongo import MongoClient
import random

client = MongoClient("mongodb://localhost:27017/")

db = client["store_db"]

products = db["products"]
```

```python
# Hapus data lama agar data sebelumnya hilang dan menjadi bersih
products.delete_many({})

# Latihan 4.1: Buatlah koleksi baru products dan masukkan data produk (minimal 10 produk).

# Daftar kategori
categories = ["Elektronik", "Fashion", "Makanan", "Minuman", "Olahraga", "Kecantikan"]

# membuat 100 produk secara acak
product_list = []
for i in range(1, 101):
    product = {
        "name": f"Produk {i}",
        "category": random.choice(categories),
        "price": random.randint(5000, 500000),
        "stock": random.randint(1, 100)
    }
    product_list.append(product)

products.insert_many(product_list)
print("100 data produk berhasil dimasukkan!")
```

```python
# Latihan 4.2: Lakukan query untuk menemukan produk dengan harga di atas rata-rata.
pipeline_avg = [
    {"$group": {"_id": None, "avg_price": {"$avg": "$price"}}}
]
avg_result = list(products.aggregate(pipeline_avg))[0]
avg_price = avg_result["avg_price"]

print(f"Rata-rata harga produk: {avg_price:.2f}")

above_avg = products.find({"price": {"$gt": avg_price}})

print("\nProduk dengan harga di atas rata-rata:")
for p in above_avg:
    print(p)
```

```python
# Latihan 4.3: Buatlah aggregation pipeline untuk menghitung total produk dalam setiap kategori.
pipeline_category = [
    {"$group": {"_id": "$category", "total_products": {"$sum": 1}}},
    {"$sort": {"total_products": -1}}
]

category_stats = list(products.aggregate(pipeline_category))

print("\nTotal produk per kategori:")
for c in category_stats:
    print(f"{c['_id']}: {c['total_products']}")
```

### 5. Tugas
- **Tugas 1**: Cari 5 karyawan dengan gaji tertinggi dalam setiap departemen, gunakan query atau aggregation yang sesuai.
- **Tugas 2**: Buatlah skenario di mana Anda harus menghapus karyawan yang berusia di bawah 25 tahun dari database.
- **Tugas 3**: Buatlah laporan ringkas (menggunakan MongoDB query) yang menghitung total gaji karyawan di setiap departemen, serta rata-rata umur karyawan.


```python
# Tugas

from pymongo import MongoClient
import random

# konek ke MongoDB lokal
client = MongoClient("mongodb://localhost:27017/")

# gunakan database dan collection
db = client["company_db"]
collection = db["employees"]
```

```python
# menghapus semua data agar bersih
collection.delete_many({})

departments = ["Finance", "IT", "HR", "Marketing", "Sales", "R&D"]

new_employees = []
for i in range(1, 51):
    emp = {
        "name": f"Karyawan_{i}",
        "department": random.choice(departments),
        "age": random.randint(22, 55),
        "salary": random.randint(3000, 10000)
    }
    new_employees.append(emp)

collection.insert_many(new_employees)
print("50 data karyawan baru berhasil ditambahkan!")
```

```python
# Tugas 1: Cari 5 karyawan dengan gaji tertinggi dalam setiap departemen, gunakan query atau aggregation yang sesuai.
pipeline_top5 = [
    {"$sort": {"department": 1, "salary": -1}},
    {"$group": {
        "_id": "$department",
        "top_employees": {"$push": {"name": "$name", "salary": "$salary"}},
    }},
    {"$project": {
        "top5": {"$slice": ["$top_employees", 5]}
    }}
]

top5_result = list(collection.aggregate(pipeline_top5))

print("5 Karyawan bergaji tertinggi per departemen:\n")
for dept in top5_result:
    print(f"Departemen: {dept['_id']}")
    for emp in dept["top5"]:
        print(f" - {emp['name']} (Gaji: {emp['salary']})")
    print()
```

```python
# Tugas 2: Buatlah skenario di mana Anda harus menghapus karyawan yang berusia di bawah 25 tahun dari database.
delete_result = collection.delete_many({"age": {"$lt": 25}})
print(f"{delete_result.deleted_count} karyawan dengan usia < 25 tahun telah dihapus.")
```

```python
# Tugas 3: Buatlah laporan ringkas (menggunakan MongoDB query) yang menghitung total gaji karyawan di setiap departemen, serta rata-rata umur karyawan.
pipeline_report = [
    {"$group": {
        "_id": "$department",
        "total_salary": {"$sum": "$salary"},
        "avg_age": {"$avg": "$age"},
        "count": {"$sum": 1}
    }},
    {"$sort": {"_id": 1}}
]

report_result = list(collection.aggregate(pipeline_report))

print("Laporan Gaji dan Umur Rata-rata per Departemen:\n")
for dept in report_result:
    print(f"Departemen: {dept['_id']}")
    print(f" - Jumlah Karyawan : {dept['count']}")
    print(f" - Total Gaji      : {dept['total_salary']}")
    print(f" - Rata-rata Umur  : {dept['avg_age']:.2f}")
    print()
```
