```python
# Exercise 1: Use bulk_write to insert, update, and delete multiple documents simultaneously.
# Task 1: Insert dataset of course where each course has multiple student enrollments, then display the data.

from pymongo import MongoClient, InsertOne, UpdateOne, DeleteOne

# Establish client connection
client = MongoClient('mongodb://localhost:27017/')
db = client['university_db']
courses_collection = db['courses']

# Bulk insert of courses with student enrollments
operations = [
    InsertOne({'course': 'Math 101', 'enrollments': 30, 'department': 'Mathematics'}),
    InsertOne({'course': 'CS 102', 'enrollments': 25, 'department': 'Computer Science'}),
    InsertOne({'course': 'History 201', 'enrollments': 20, 'department': 'History'}),
    InsertOne({'course': 'Physics 202', 'enrollments': 15, 'department': 'Physics'})
]
courses_collection.bulk_write(operations)
print('Courses inserted successfully.')
```

```python
for db_name in client.list_database_names():
    if db_name not in ['admin', 'local', 'config']:
        print(f"Deleted: {db_name}")
        client.drop_database(db_name)

print('All Data Clean Up')
```

```python
client = MongoClient('mongodb://localhost:27017/')
db = client['university_db']
courses_collection = db['courses']

courses_collection.delete_many({})

operations = [
    InsertOne({
        'course': 'Math 101',
        'department': 'Mathematics',
        'enrollments': [
            {'student_id': 1, 'name': 'Alice', 'grade': 'A'},
            {'student_id': 2, 'name': 'Bob', 'grade': 'B+'},
            {'student_id': 3, 'name': 'Charlie', 'grade': 'A-'}
        ]
    }),
    InsertOne({
        'course': 'CS 102',
        'department': 'Computer Science',
        'enrollments': [
            {'student_id': 4, 'name': 'Diana', 'grade': 'A'},
            {'student_id': 5, 'name': 'Evan', 'grade': 'B'},
            {'student_id': 6, 'name': 'Frank', 'grade': 'A'}
        ]
    }),
    InsertOne({
        'course': 'History 201',
        'department': 'History',
        'enrollments': [
            {'student_id': 7, 'name': 'Grace', 'grade': 'B+'},
            {'student_id': 8, 'name': 'Hank', 'grade': 'A-'},
            {'student_id': 9, 'name': 'Irene', 'grade': 'A'}
        ]
    })
]

courses_collection.bulk_write(operations)
print("Dataset courses berhasil dimasukkan.\n")

print("Daftar Mata Kuliah dan Enrollments")
for course in courses_collection.find():
    print(f"\nCourse: {course['course']} ({course['department']})")
    print("Enrollments:")
    for student in course['enrollments']:
        print(f" - {student['name']} (ID: {student['student_id']}), Grade: {student['grade']}")
```

```python
# regular expression (regex)
print("Courses with regex (not case sensitive)")
departments = ['computer science', 'mathematics']
regex_filters = [{'department': {'$regex': dept, '$options': 'i'}} for dept in departments]

for course in courses_collection.find({'$or': regex_filters}):
    print(course)
```

```python
# Exercise 3: Use agregation to find the average enrollment per departments
pipeline = [
    {'$group': {'_id': '$department', 'average_enrollment': {'$avg': '$enrollments'}}}
]
for result in courses_collection.aggregate(pipeline):
    print(result)
```

```python
# Task 3: Create an aggregation pipeline that finds the maximum enrollment for each department.

pipeline = [
    {'$group': {'_id': '$department', 'average_enrollment': {'$max': '$enrollments'}}}
]
for result in courses_collection.aggregate(pipeline):
    print(result)
```

```python
# menghitung rata-rata grade per departemen

pipeline = [
    {'$unwind': '$enrollments'},
    {'$addFields': {
        'numeric_grade': {
            '$switch': {
                'branches': [
                    {'case': {'$eq': ['$enrollments.grade', 'A']}, 'then': 4.0},
                    {'case': {'$eq': ['$enrollments.grade', 'A-']}, 'then': 3.7},
                    {'case': {'$eq': ['$enrollments.grade', 'B+']}, 'then': 3.3},
                    {'case': {'$eq': ['$enrollments.grade', 'B']}, 'then': 3.0}
                ],
                'default': 0
            }
        }
    }},
    {'$group': {'_id': '$department', 'avg_grade': {'$avg': '$numeric_grade'}}}
]

for result in courses_collection.aggregate(pipeline):
    print(result)
```

```python
# Exercise 4: Use project to rename and only show fields: course_name, department_name, and enrollments.

pipeline = [
    {'$project': {'course_name': '$course', 'department_name': '$department', 'enrollments': 1}}
]
for result in courses_collection.aggregate(pipeline):
    print(result)
```

```python
# Task 4: Use $addFields to create a new field enrollment_category where enrollments > are 'high' and others 'low'
pipeline = [
    {'$addFields': {'enrollment_category': {'$cond': {'if': {'$gt': ['$enrollments', 20]}, 'then': 'high', 'else': 'low'}}}}
]
for result in courses_collection.aggregate(pipeline):
    print(result)
```
