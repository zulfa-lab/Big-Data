```python
# Exercise 1: Identifying and Handling Missing Data
import pandas as pd

# Sample dataset with missing values
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', None],
    'Age': [24, 30, None, 22, 35],
    'Salary': [48000, None, 57000, None, 6000]
}
df = pd.DataFrame(data)
print('Before cleaning:\n', df)

# Filling missing values and droppping rows
df['Age'].fillna(df['Age'].mean(), inplace=True)
df['Salary'].fillna(df['Salary'].median(), inplace=True)
df.dropna(subset=['Name'], inplace=True)
print('\nAfter cleaning:\n', df)
```

```python
# Exercise 2: Standarizing Categorical Data
# Sample dataset with inconsistent categorical values
data = {
    'Product': ['Laptop', 'Laptop', 'Desktop', 'Tablet', 'Tablet'],
    'Category': ['Electronics', 'electronics', 'Electronics', 'Gadgets', 'gadgets']
}
df = pd.DataFrame(data)

# Standarize category values
df['Category'] = df['Category'].str.capitalize()
print('Standardized Data:\n', df)
```

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
pd.set_option('display.max_columns', None)
# muat dataset
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

print("1. Data mentah (10 Baris Pertama)")
print(df.head(10))
print("\n" + "="*40 + "\n")

# mengidentifikasi nilai yang hilang
print("2. Nilai yang hilang")
print(df.isnull().sum())
print("\n" + "="*40 + "\n")

#mengisi nilai umur berdasarkan rata-rata
df['Age'].fillna(df['Age'].mean(), inplace=True)
mean = df.Age.mean()
print("Rata-rata Age:", mean)
# mengisi 'Embarked' dengan modus
df['Embarked'].fillna(df['Embarked'].mode()[0], inplace=True)
modus = df.Embarked.mode()[0]
print("Modus Embarked:", modus)
# menghapus kolom kabin yang tidak ada isinya
df.drop('Cabin', axis=1, inplace=True)

print("2b. Mengecek nilai yang hilang setelah imputasi")
print(df.isnull().sum())
print("\n" + "="*40 + "\n")

# menghapus duplikasi
ttl_brs_masih_duplikasi = len(df)
df.drop_duplicates(inplace=True)
ttl_brs_stlah_hps_dplksi = len(df)

print(f"3. Hapus Duplikat")
print(f"Jumlah baris awal: {ttl_brs_masih_duplikasi}")
print(f"Jumlah baris setelah menghapus duplikasi: {ttl_brs_stlah_hps_dplksi}")
print(f"Total duplikat dihapus: {ttl_brs_masih_duplikasi - ttl_brs_stlah_hps_dplksi}")
print("\n" + "="*40 + "\n")

# normalisasi kolom angka (numerik)
print("4. Normalisasi Kolom Numerik ('Age' & 'Fare')")
kolom_numerik = ['Age', 'Fare']

scaler = MinMaxScaler()

df[kolom_numerik] = scaler.fit_transform(df[kolom_numerik])

print("Data 'Age' dan 'Fare' setelah normalisasi:")
print(df[['Age', 'Fare']].head())
print("\nDeskripsi data 'Age' dan 'Fare' setelah normalisasi:")
print(df[['Age', 'Fare']].describe())
print("\n" + "="*40 + "\n")

#standarasisasi kolom kategori
print("5. Standarisasi kolom kategori ('Sex' & 'Embarked')")

# Kolom yang akan di-encode
kolom_kategori = ['Sex', 'Embarked', 'Pclass']

df_bersih = pd.get_dummies(df, columns=kolom_kategori, drop_first=True)

print("Data final setelah  di-encoding (5 Baris Pertama):")
print(df_bersih.head())
print("\nKolom-kolom baru yang terbentuk:")
print(df_bersih.columns)
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler

url = "https://archive.ics.uci.edu/ml/machine-learning-databases/heart-disease/processed.cleveland.data"

nama_kolom = ['age', 'sex', 'cp', 'trestbps', 'chol', 'fbs', 'restecg',
                'thalach', 'exang', 'oldpeak', 'slope', 'ca', 'thal', 'target']

df = pd.read_csv(url, header=None, names=nama_kolom)

print("1. Data mentah (5 Baris Pertama)")
print(df.head())
print("\nInfo awal (perhatikan tipe data 'object' yang aneh):")
print(df.info())
print("\n" + "="*40 + "\n")


# 2. Pembersihan data dan mengisi nilai yang hilang
print(f"Baris dengan '?' sebelum diganti: {len(df[df.eq('?').any(axis=1)])}")

# mengganti ? dengan nan
df.replace('?', np.nan, inplace=True)

print("\nJumlah missing values (NaN) setelah '?' diganti:")
print(df.isnull().sum())

df['ca'] = pd.to_numeric(df['ca'])
df['thal'] = pd.to_numeric(df['thal'])

# Imputasi (isi) dengan mean (rata-rata)
df['ca'].fillna(df['ca'].mean(), inplace=True)
df['thal'].fillna(df['thal'].mean(), inplace=True)

print("\nMissing values setelah diimputasi:")
print(df.isnull().sum())
print("\n" + "="*40 + "\n")


# 3. Penanganan outlier
kolom_outlier = ['trestbps', 'chol', 'thalach']
df_bersih = df.copy()

print(f"3. Penanganan Outliers")
print(f"Jumlah baris sebelum outlier dihapus: {len(df_bersih)}")

for col in kolom_outlier:
    Q1 = df_bersih[col].quantile(0.25)
    Q3 = df_bersih[col].quantile(0.75)
    IQR = Q3 - Q1

    nilai_terkecil = Q1 - 1.5 * IQR
    nilai_terbesar = Q3 + 1.5 * IQR


    df_bersih = df_bersih[(df_bersih[col] >= nilai_terkecil) & (df_bersih[col] <= nilai_terbesar)]

print(f"Jumlah baris setelah outlier dihapus: {len(df_bersih)}")
print("\n" + "="*40 + "\n")


# 4. Normalisasi
kolom_numerik = ['age', 'trestbps', 'chol', 'thalach', 'oldpeak']

scaler = MinMaxScaler()

df_bersih[kolom_numerik] = scaler.fit_transform(df_bersih[kolom_numerik])

print("4. Data Setelah Normalisasi (5 Baris Pertama)")
print(df_bersih.head())
print("\nDeskripsi data numerik setelah normalisasi (min=0, max=1):")
print(df_bersih[kolom_numerik].describe())
print("\n" + "="*40 + "\n")


# 5. One hot encoding
kolom_kategori = ['sex', 'cp', 'fbs', 'restecg', 'exang', 'slope', 'ca', 'thal']

df_final = pd.get_dummies(df_bersih, columns=kolom_kategori, drop_first=True)

print("5. Data Final Siap untuk Analisis (5 Baris Pertama)")
print(df_final.head())
print(f"\nJumlah kolom akhir: {len(df_final.columns)}")
```
